3. Write a Python program to:

a. Create a dictionary to store key->Roll number and value-> student name:
b. Ask the user to input a new roll number and student name.
c. If the roll number already exists, print "Roll number already registered".
d. Otherwise, add the student to the dictionary.
e. Display the updated dictionary.

/////////////////////////////////////////////////////////////

```python
# Function to create initial dictionary
def create_students():
    return {
        "101": "Ali",
        "102": "Ahmed"
    }


# Function to add new student                          students = {}
def add_student(students):
    roll = input("Enter new roll number: ")
    name = input("Enter student name: ")

    if roll in students:
        print("Roll number already registered")
    else:
        students[roll] = name
        print("Student added successfully")


# Function to display dictionary
def display_students(students):
    print("\nUpdated Student Dictionary:")
    for r, n in students.items():
        print(r, ":", n)                #print(f"{roll} -> {name}") r=roll,n=name


# Main Program 
students = create_students()

while True:
    print("\n------ MENU ------")
    print("1. Add Student")
    print("2. View Student List")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_student(students)
    elif choice == "2":
        display_students(students)
    elif choice == "3":
        print("Exiting program...")
        break
    else:
        print("Invalid choice! Try again.")
```

2. Write a Python program to compare readings from 5 sensors.

a. Store readings in a dictionary.
b. Input the readings from the user.
c. Determine the difference between the highest and lowest readings

/////////////////////////////////////////////////////////////

```python
# Function to input sensor readings
def input_sensors():
    sensors = {}
    for i in range(1, 6):
        name = input(f"Enter sensor {i} name: ")
        reading = float(input(f"Enter reading of {name}: "))
        sensors[name] = reading
    return sensors


# Function to calculate difference
def calculate_difference(sensors):
    highest = max(sensors.values())
    lowest = min(sensors.values())
    return highest - lowest


# Main execution
sensors = input_sensors()
difference = calculate_difference(sensors)

print("\nDifference between highest and lowest readings:", difference)
```

1. Write a Python program that records readings from 5 smart devices.

a. Store the device readings in a dictionary.
b. Accept user input for readings.
c. Sort the sensors in descending order based on their readings.
d. Display the top 2 highest-reading sensors.
e. Identify which sensor has the lowest reading and display its name.

/////////////////////////////////////////////////////////////

```python

# Function to input device readings
def input_devices():
    devices = {}
    for i in range(1, 6):
        name = input(f"Enter device {i} name: ")
        reading = float(input(f"Enter reading of {name}: "))
        devices[name] = reading
    return devices


# Function to sort devices in descending order
def sort_devices(devices):
    device_list = list(devices.items())   # convert dictionary to list
    
    device_list.sort(key=lambda x: x[1], reverse=True)   # descending
    
    return device_list


# Function to display results
def display_results(devices):
    sorted_devices = sort_devices(devices)

    print("\nDevices Sorted (Descending Order):")
    for device, reading in sorted_devices:
        print(device, ":", reading)

    print("\nTop 2 Highest-Reading Devices:")
    for i in range(2):
        print(sorted_devices[i][0], ":", sorted_devices[i][1])

    lowest_device = min(devices, key=devices.get)
    
    print("\nLowest Reading Device:")
    
    print(lowest_device, ":", devices[lowest_device])


# Main Program
devices = input_devices()
display_results(devices)
```
