2 number input,  perform arthematic opertaion in flot and int and print 

```python
a=int(input("Enter a 1st number"))
b=int(input("Enter a 2nd number"))
print("first number is: ",a)
print("Second number is:",b)


print('\n','int type')
a=int(a)
b=int(b)
x1=a+b
x2=a**b
x3=a//b
x4=a-b

print(x1)
print(x2)
print(x3)
print(x4)

print('\n','float type')
a=float(a)
b=float(b)
x1=a+b
x2=a*b
x3=a/b
x4=a-b

print(x1)
print(x2)
print(x3)
print(x4)


```
