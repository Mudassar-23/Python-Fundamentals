Immutable Data Types:
Cannot be changed after creation
(If you modify them, a new object is created.)

Types:

int → 10
float → 3.14
complex → 2 + 3j
bool → True / False
str (String) → "Hello"
tuple → (1, 2, 3)


x = 10
x = x + 5   # New object created

s = "Hello"
s[0] = "h"   #  Error (String is immutable)

2. Mutable Data Types

 Can be changed after creation

Types:
list → [1, 2, 3]
dict (Dictionary) → {"name": "Ali"}


List (Important Mutable Type)
A list is an ordered collection of items.

Example:
numbers = [1, 2, 3]
numbers.append(4)   # Modify list
print(numbers)

✔ Output:
[1, 2, 3, 4]

You can:
Add items → append()
Remove items → remove()
Change items → numbers[0] = 10
